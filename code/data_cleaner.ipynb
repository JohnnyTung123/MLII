```python
import pandas as pd
import numpy as np
pd.set_option('display.max_rows', False)
```

```python
path = '../data/raw/usa characteristics.csv'
data = pd.read_csv(path)
data
```

```python
print(data.columns.tolist())
```

```python
id_unique = sorted(data['id'].unique())
print(id_unique)
```

```python
# 'market_equity', 'ret_exc', 'ret_local', 'me' contains forward information
# data = data.drop(columns=['market_equity', 'ret_exc', 'ret_local', 'me'])
data = data.drop(columns=['market_equity', 'ret_exc', 'ret_local', 'me', "ret_1_0", "ret"])
```

```python
# size grp: we probably don't want micro and nano cap because they are illiquid and hence incur high t-cost or artificial high return
print(data['size_grp'].unique())
data = data[~data['size_grp'].isin(['nano', 'micro', np.nan])]
```

```python
# if CRSP exists, use CRSP as the main observation.
data = data[data['obs_main'] == 1]
# only regular common shares
data = data[data['common'] == 1]
# only includes stocks listed on the primary/main exchange.
data = data[data['exch_main'] == 1]
# non-primary listings
data = data[data['primary_sec'] == 1]
data
```

```python
# factor_df['date'] = pd.to_datetime(factor_df['date'].astype(str), format='%Y%m%d')
# factor_df['date'] = factor_df['date'].apply(lambda x: x.strftime('%Y%m%d'))
# factor_df['date'] = factor_df['date'].astype(int)
data['date'] = pd.to_datetime(data['date'], format='%Y%m%d')
start_dt = data['date'].min()
end_dt = data['date'].max()
print ("Date Range: %s --> %s"%(start_dt,end_dt))
date_unique = sorted(data['date'].unique())
```

```python
data['eom'] = pd.to_datetime(data['eom'], format='%Y%m%d')
start_eom = data['eom'].min()
end_eom = data['eom'].max()
print ("EOM Range: %s --> %s"%(start_eom,end_eom))
eom_unique = sorted(data['eom'].unique())
```

```python
for eom in eom_unique:
    print(eom)
```

```python
data = data.sort_values(by=['id', 'eom'])
data
```

```python
# use factors_t to predict return_t+1
data[['id', 'eom', 'ret_exc_lead1m']] # Excess return (ret exc*) in month t+1
```
